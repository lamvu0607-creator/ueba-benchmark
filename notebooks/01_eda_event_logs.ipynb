# UEBA Benchmark - EDA Windows Event Logs (4624 & 4625)

Notebook này phục vụ công việc khám phá dữ liệu (EDA) tuần đầu:
1. Phân phối thời gian đăng nhập (Giờ trong ngày, ngày trong tuần)
2. Tỷ lệ thành công / thất bại (`4624` vs `4625`)
3. Phân phối các kiểu đăng nhập (`LogonType` 2: Interactive, 3: Network, 10: RDP)
4. Nhận diện các tài khoản hoạt động nhiều nhất (Top active users)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
# Load dữ liệu interim đã chuẩn hóa
interim_path = Path("../data/interim/cleaned_logs.parquet")
if interim_path.exists():
    df = pd.read_parquet(interim_path)
    print(f"Loaded {len(df)} records.")
    display(df.head())
else:
    print("Chưa có file interim. Hãy chạy `python main.py --stage clean` trước.")
```

```python
# Phân phối Event ID 4624 vs 4625
if 'df' in locals():
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='event_id')
    plt.title("Tỷ lệ Event 4624 (Success) vs 4625 (Failure)")
    plt.show()
```
